```python
import pandas as pd
import numpy as np
```

```python
def resample_sesnor(data,start,end,newSamplingRate):
    newX = np.arange(start,end,1000/newSamplingRate).astype(int)
    merged = np.append(data.index.values,newX)
    merged.sort()
    merged = np.unique(merged)
    data = data.reindex(merged)
    data = data.interpolate()
    data = data.reindex(newX)
    return data
```

```python
all_data = {}

for p in ['P0','P1', 'P2', 'P3', 'P4', 'P5', 'P6', 'P7']:
    print(p)
    data = pd.read_csv("../Data/Pilot/Raw/"+p+"/MLX.csv",index_col=1,header=None)
    data = data.drop([0],axis=1)
    data = data.dropna(how='all',axis=0)
    data = data.dropna(how='all',axis=1)
    
    all_data[p] = data.copy()
    
all_data['P0'] = all_data['P0'].iloc[41:]    
```

```python
for p in ['P0','P1', 'P2', 'P3', 'P4', 'P5', 'P6', 'P7']:
    print(p)
    data = all_data[p].copy()
    resampled = resample_sesnor(data,data.index[0],data.index[-1],4)
    resampled.to_csv("../Data/Pilot/Clean/"+p+"/MLX.csv",header=False)
```
